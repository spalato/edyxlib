```python
# This is a template for processing linear absorption spectra. Uncomment as necessary.
import numpy as np
import matplotlib.pyplot as plt
from glob import glob # glob is a standard python module that helps select multiple files

# this is new
from pyspan import Linear
```

```python
# this is necessary. It makes jupyter use ipywidgets for matplotlib plots
%matplotlib widget 
```

# Set list of files

```python
# get file names, somehow.
fnames = glob('s[1][4].dat')#glob("s02.dat")+glob("s32.dat") #"s[0][0-1].dat"

date = '250307_'
sample = 'LCH' 

print(fnames)
```

# Inspect the data

```python
l = Linear.from_files(fnames)
l.plot()
```

# Reject bad spectra and average over delays 

```python
# get average spectra over delays as dict
mean_spectra = l.spectra_dict
```

```python
for fn, (wl, od) in mean_spectra.items():
    plt.figure()
    plt.plot(wl, od, label=fn)
    plt.xlabel(r"$\lambda$/ nm") 
    plt.ylabel("A/ mOD") 
    plt.axvline(526, color='r', linewidth=0.5)
    plt.title(fn)
    plt.grid()
    plt.savefig(date+fn[:-4]+".png", bbox_inches='tight', dpi=300) 
```
